```python
import pandas as pd
import numpy as np
import statistics as st
df =pd.read_csv("/content/Dataset_SummaryStatistics_University.csv")
df
```

```python
df.describe() #Provides summary measures for variables of type : int or float
```

```python
print('Mean ', st.mean(df['GPA']))    #prints mean of only GPA Variable
print('Median ', st.median(df['GPA']))    #prints median of only GPA Variable
print('Mode ', st.mode(df['GPA']))    #prints mode of only GPA Variable
print('Standard Deviation ', st.pstdev(df['GPA']))    #prints population standard deviation
print('Standard Deviation ', st.stdev(df['GPA']))    #prints sample standard deviation
```
