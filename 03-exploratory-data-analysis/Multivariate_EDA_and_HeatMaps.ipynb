```python
import seaborn as sns
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Dataset_Scatterplot_HeatMap_Histogram_pairplot.csv')
df.head()
```

```python
#GPA vs Study Hours: Should show a positive trend (more hours → higher GPA)
#Study Hours vs Attendance: May or may not show a clear relationship



# Plot 1: GPA vs Study Hours
plt.figure(figsize=(6, 4))
sns.scatterplot(x='Study_Hours', y='GPA', data=df, color='blue')
plt.title("GPA vs Study Hours")
plt.xlabel("Study Hours")
plt.ylabel("GPA")
plt.tight_layout()
plt.show()
```

```python
# GPA vs Attendance: Also positively correlated

plt.figure(figsize=(6, 4))
sns.scatterplot(x='Attendance', y='GPA', data=df, color='green')
plt.title("GPA vs Attendance")
plt.xlabel("Attendance(%)")
plt.ylabel("GPA")
plt.tight_layout()
plt.show()
```

```python
#Study Hours vs Attendance: May or may not show a clear relationship
plt.figure(figsize=(6, 4))
sns.scatterplot(x='Attendance', y='Study_Hours', data=df, color='orange')
plt.title("Study Hours vs Attendance")
plt.xlabel("Attendance(%)")
plt.ylabel("Study Hours")
plt.tight_layout()
plt.show()
```

```python
#Pairplot
sns.set(style="ticks")
sns.pairplot(df, diag_kind="kde", plot_kws={"alpha": 0.7, "s": 60, "edgecolor": "k"})

# Show the plot
plt.suptitle("Pair Plot: GPA, Study Hours, Attendance", y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Linear Regression
# Set style
sns.set(style="whitegrid")

# Plot 1: GPA vs Study Hours
plt.figure(figsize=(6, 4))
sns.regplot(x='Study_Hours', y='GPA', data=df, scatter_kws={'s': 50, 'alpha': 0.7}, line_kws={'color': 'red'})
plt.title("GPA vs Study Hours (with Trendline)")
plt.xlabel("Study Hours")
plt.ylabel("GPA")
plt.tight_layout()
plt.show()
```

```python
# Compute the correlation matrix
corr_matrix = df.corr(numeric_only=True)

# Set style and figure size
plt.figure(figsize=(8, 6))
sns.set(font_scale=1.1)

# Draw the heatmap
sns.heatmap(
    corr_matrix,
    annot=True,            # Show the correlation coefficients
    cmap='coolwarm',       # Color map
    fmt=".2f",             # Format float to 2 decimal places
    linewidths=0.5,        # Line between cells
    square=True,           # Make cells square-shaped
    cbar_kws={"shrink": .8} # Shrink color bar
)

plt.title("Correlation Heatmap: GPA, Study Hours, Attendance Rate")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.histplot(df['GPA'], bins=10, kde=True, color='skyblue')
plt.title("Distribution of GPA")
plt.xlabel("GPA")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
# Plot: Histogram of Study Hours
plt.figure(figsize=(6, 4))
sns.histplot(df['Study_Hours'], bins=10, kde=True, color='orange')
plt.title("Distribution of Study Hours")
plt.xlabel("Study Hours")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
# Plot: Histogram of Attendance Rate
plt.figure(figsize=(6, 4))
sns.histplot(df['Attendance'], bins=10, kde=True, color='green')
plt.title("Distribution of Attendance")
plt.xlabel("Attendance (%)")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```
