```python
import matplotlib.pyplot as plt

# Daily temperatures with outliers
temperatures = [22, 24, 19, 21, 18, 25, 30, 28, 27, 23, 26, 29, 24, 22, 20, 31, 33, 28, 26, 25, 45, 5]

plt.boxplot(temperatures, vert=True, whis=1.5, showfliers=True,meanline = True) #default whisker value
plt.show()

plt.boxplot(temperatures, vert=True, whis=2.0, showfliers=True) # whisker = 2 fewer outlier
plt.show()

# Boxplot with whiskers covering full data range
range=[min(temperatures),max(temperatures)]
plt.boxplot(temperatures, vert=True, whis=range, showfliers=True)

plt.title("Box Plot of Daily Temperatures (°C)")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
plt.boxplot(temperatures, meanline=True, showmeans=True, vert=True)   #to show the mean of the datapoints
plt.show()
```

```python
df = pd.read_csv("/content/Dataset_BarBoxplot_Train.csv")

 # Calculate IQR and detect outliers
Q1 = df['ApplicantIncome'].quantile(0.25)
Q3 = df['ApplicantIncome'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['ApplicantIncome'] < lower_bound) | (df['ApplicantIncome'] > upper_bound)]
print("Outliers in 'ApplicantIncome' column:")
print(outliers)

#  Box plot
plt.boxplot(df['ApplicantIncome'], vert=True, whis=1.5, showfliers=True)
plt.title("Box Plot of ApplicantIncome")
plt.ylabel("ApplicantIncome")
plt.show()
```
