```python

#Scenario:
#A startup wants to study whether marketing spend predicts sales revenue. You have data for 50 campaigns. Your task is to:

#1.Plot a scatterplot to check linear relationship

#2.If linear, calculate the Pearson correlation coefficient

#3.If correlation is > 0.7, perform Simple Linear Regression

#4.Visualize the regression line
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import linregress, pearsonr
```

```python

# Generate synthetic data50
np.random.seed(0)
x = np.random.normal(50, 10, 50)  # Marketing Spend
y = 3 * x + np.random.normal(0, 1, 50)  # Sales with some noise
```

```python
# Scatterplot
plt.figure(figsize=(8, 5))
sns.scatterplot(x=x, y=y)
plt.title("Marketing Spend vs Sales Revenue")
plt.xlabel("Marketing Spend (in $1000s)")
plt.ylabel("Sales Revenue (in $1000s)")
plt.grid(True)
plt.show()
```

```python
# Pearson correlation
corr, _ = pearsonr(x, y)
print(f"Pearson Correlation Coefficient: {round(corr, 3)}")
```

```python
# Simple Linear Regression only if corr > 0.7
if corr > 0.7:
    slope, intercept, r_value, p_value, std_err = linregress(x, y)
    print(f"Regression Line: y = {round(slope, 2)}x + {round(intercept, 2)}")
    print(f"R² = {round(r_value**2, 3)}")
```

```python
# Safe version to avoid plotting errors
x_sorted = np.sort(x)
y_pred_sorted = slope * x_sorted + intercept

plt.figure(figsize=(8, 5))
plt.scatter(x, y, label='Actual Data')
plt.plot(x_sorted, y_pred_sorted, color='red', label='Regression Line')
plt.xlabel("Marketing Spend (in $1000s)")
plt.ylabel("Sales Revenue (in $1000s)")
plt.title("Simple Linear Regression: Spend vs Sales")
plt.legend()
plt.grid(True)
plt.show()
```
