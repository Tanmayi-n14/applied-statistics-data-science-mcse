```python
'''
Scenario:A shoe manufacturer claims that the average lifespan of their sports shoes is 400 days. A market research agency collects a large sample of 200 customer reviews, and the sample mean lifespan is found to be 395 days with a standard deviation of 20 days.
You're asked to:
1.Perform a two-tailed Z-test to check if the actual mean differs from the claimed value.
2.Use α = 0.05
3.Calculate the p-value
4.Draw a conclusion about the claim
'''
```

```python
#Python Code (Z-Test for Large Sample)
import numpy as np
from scipy.stats import norm

```

```python
# Given
sample_mean = 395
pop_mean = 400
std_dev = 20
n = 200
alpha = 0.05
```

```python


# Z-test statistic
z_score = (sample_mean - pop_mean) / (std_dev / np.sqrt(n))
p_value = 2*(1-norm.cdf(abs(z_score)))

# 4.Draw a conclusion about the claie = 2 * (1 - norm.cdf(abs(z_score)))  # two-tailed

print("Z-Score:", round(z_score, 3))
print("P-Value:", round(p_value, 4))

if p_value < alpha:
    print(" Reject the null hypothesis: Lifespan is significantly different.")
else:
    print("Fail to reject the null hypothesis: No significant difference.")
```
