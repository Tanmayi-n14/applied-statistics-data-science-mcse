```python
from bs4 import BeautifulSoup as bs
import requests
from fake-useragent import UserAgent

ua = UserAgent()
header = { 'User-Agent':str(ua.chrome)}

link='https://www.scrapethissite.com/pages/simple/'
page=requests.get(link, headers=header)
print(page)
```

```python
soup=bs(page.content,'html.parser')
#print(soup)  //raw html data

print(soup.prettify())
```

```python
name=soup.find_all("h3",class_="country-name")
print(name)
```

```python
cust_name=[]
for i in range(0,len(name)):
    cust_name.append(name[i].get_text())
cust_name
print(cust_name)
len(cust_name)
```

```python
capital = soup.find_all("span", class_="country-capital")
population = soup.find_all("span", class_="country-population")
area = soup.find_all("span", class_="country-area")
```

```python
cust_capital = [cap.get_text() for cap in capital]
cust_population = [pop.get_text() for pop in population]
cust_area = [ar.get_text() for ar in area]
```

```python
import pandas as pd
df=pd.DataFrame()
df['Customer Name']=cust_name
df['Capital']=cust_capital
df['Population']=cust_population
df['Area']=cust_area
display(df.head())
```

```python
df.to_csv('abc.csv', index=False, encoding='utf-8')
```
