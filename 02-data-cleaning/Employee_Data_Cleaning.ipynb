```python
df1 = pd.DataFrame({
    'brand': ['Maggi', 'Maggi', 'Yippee', 'Yippee', 'Top Ramen'],
    'style': ['pack', 'pack', 'cup', 'pack', 'pack'],
    'rating': [4.5, 4.5, 3.8, 14, 5]
})
df1
```

```python
df1.drop_duplicates() # removes the duplicated rows
```

```python
#To remove duplicates on specific column(s), use subset.
df1.drop_duplicates(subset=['brand'])
```

```python
#To remove duplicates on specific column(s), use subset.
#to remove duplicates and keep last occurrences, use keep.
df1.drop_duplicates(subset=['brand', 'style'], keep='last')
```

```python
#Categorical column missing values replaced with mode
#df.Employment_Type.mode()[0]
df.Employment_Type.fillna(df.Employment_Type.mode()[0])
```

```python
df.Age.median()
```

```python
#numerical discrete values replaced with median.
df.Age.fillna(df.Age.median())
```

```python
# Numerical Continuous - replace missing values with mean
df['Salary'].fillna(df['Salary'].mean(),inplace=True)
```
